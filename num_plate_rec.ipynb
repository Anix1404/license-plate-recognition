```python
import cv2
import easyocr
import matplotlib.pyplot as plt
from ultralytics import YOLO

# 1. Load your trained YOLOv8 model
model = YOLO("/content/best (1).pt")

# 2. Initialize EasyOCR reader
reader = easyocr.Reader(["en"], gpu=True)

# 3. Load the test image
image_path = "/content/120260519165511UP93CT6900.jpg"
image = cv2.imread(image_path)

if image is None:
  print(f"Error: Could not load image from {image_path}. Check the path!")
else:
  # 4. Run YOLOv8 detection
  results = model(image)

  # 5. Crop the license plate and run OCR
  for r in results:
    boxes = r.boxes.xyxy
    for box in boxes:
      x1, y1, x2, y2 = map(int, box[:4])
      plate_crop = image[y1:y2, x1:x2]

      # Optional: Show the cropped plate to check its quality
      plt.figure(figsize=(4, 2))
      plt.imshow(cv2.cvtColor(plate_crop, cv2.COLOR_BGR2RGB))
      plt.axis("off")
      plt.title("Detected License Plate Crop")
      plt.show()

      # Run EasyOCR
      ocr_results = reader.readtext(plate_crop)

      # Combine all text pieces found on the plate and remove spaces
      raw_text = "".join([detection[1] for detection in ocr_results])
      cleaned_plate_text = "".join(raw_text.split()).upper()

      print(f"Final Cleaned Plate Text: {cleaned_plate_text}")
```

```python
import os
import zipfile
from ultralytics import YOLO

# 1. Unzip your dataset (Replace with your exact zip filename if different)
zip_filename = "/content/license_plate_detection.v1i.yolov8.zip"  # adjust if needed
extract_path = "/content/dataset"

if os.path.exists(zip_filename):
  print("Unzipping dataset...")
  with zipfile.ZipFile(zip_filename, "r") as zip_ref:
    zip_ref.extractall(extract_path)
  print("✅ Dataset extracted successfully!")
else:
  print(
      f"❌ Could not find {zip_filename}. Make sure it's uploaded to /content/"
  )

# 2. Load your trained YOLOv8 model
model = YOLO("/content/best (1).pt")

# 3. Find the data.yaml path automatically and run validation
yaml_path = None
for root, dirs, files in os.walk(extract_path):
  if "data.yaml" in files:
    yaml_path = os.path.join(root, "data.yaml")
    break

if yaml_path:
  print(f"Found config at: {yaml_path}")
  print("Running validation to calculate accuracy metrics...")

  # Run YOLOv8 validation mode
  metrics = model.val(data=yaml_path)

  # Print accuracy results
  print("\n--- Model Validation Results ---")
  print(f"mAP50 (Accuracy at 0.5 IoU): {metrics.box.map50:.4f}")
  print(f"mAP50-95 (Overall Accuracy): {metrics.box.map:.4f}")
  print(f"Precision: {metrics.box.p.mean():.4f}")
  print(f"Recall: {metrics.box.r.mean():.4f}")
else:
  print("❌ 'data.yaml' not found inside the extracted folder.")
```

```python
import cv2
import easyocr
from ultralytics import YOLO

# 1. Load trained YOLOv8 model and initialize EasyOCR
model = YOLO("weights/best.pt")
reader = easyocr.Reader(["en"], gpu=True)

# 2. Connect to RTSP Stream (replace with your camera URL)
rtsp_url = "rtsp://username:password@ip_address:554/stream_path"
cap = cv2.VideoCapture(rtsp_url)

if not cap.isOpened():
  print("Error: Could not open RTSP stream. Check URL and network.")
  exit()

print("Streaming active. Press 'q' to exit.")

while cap.isOpened():
  ret, frame = cap.read()
  if not ret:
    print("Stream disconnected or frame empty. Reconnecting...")
    break

  # Run YOLOv8 detection on the live frame
  results = model(frame, verbose=False)

  for r in results:
    boxes = r.boxes.xyxy
    for box in boxes:
      x1, y1, x2, y2 = map(int, box[:4])

      # Ensure coordinates stay within frame bounds
      h, w, _ = frame.shape
      x1, y1 = max(0, x1), max(0, y1)
      x2, y2 = min(w, x2), min(h, y2)

      # Crop the detected license plate region
      plate_crop = frame[y1:y2, x1:x2]
      if plate_crop.size == 0:
        continue

      # Run EasyOCR and post-process text
      ocr_results = reader.readtext(plate_crop)
      raw_text = "".join([detection[1] for detection in ocr_results])
      cleaned_text = "".join(raw_text.split()).upper().replace("CI", "CT")

      # Draw visual box and extracted text overlay on the stream
      cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
      cv2.putText(
          frame,
          cleaned_text,
          (x1, y1 - 10),
          cv2.FONT_HERSHEY_SIMPLEX,
          0.9,
          (0, 255, 0),
          2,
      )

      if cleaned_text:
        print(f"Detected Plate: {cleaned_text}")

  # Show live video window with annotations
  cv2.imshow("Live License Plate Recognition", frame)
  if cv2.waitKey(1) & 0xFF == ord("q"):
    break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import easyocr
from ultralytics import YOLO

# 1. Load your trained YOLOv8 model weights
model = YOLO("/content/best (1).pt")

# 2. Initialize EasyOCR reader for English characters/numbers
reader = easyocr.Reader(["en"], gpu=True)

# 3. Path to the test image you want to evaluate manually
image_path = (
    "/content/120260519165511UP93CT6900.jpg"  # Change to your image path
)
image = cv2.imread(image_path)

if image is None:
  print(f"Error: Could not load image from {image_path}. Check the file path!")
else:
  # 4. Run YOLOv8 detection
  results = model(image)

  # 5. Crop the detected license plate and run OCR
  for r in results:
    boxes = r.boxes.xyxy
    for box in boxes:
      x1, y1, x2, y2 = map(int, box[:4])
      plate_crop = image[y1:y2, x1:x2]

      # Run EasyOCR on the cropped plate region
      ocr_results = reader.readtext(plate_crop)

      # Combine text segments and clean spacing
      raw_text = "".join([detection[1] for detection in ocr_results])
      cleaned_plate_text = "".join(raw_text.split()).upper()

      # Post-processing fix for common OCR misreads (e.g., T read as I)
      cleaned_plate_text = cleaned_plate_text.replace("CI", "CT")

      print(
          f"Successfully Extracted Plate Number: {cleaned_plate_text}"
      )
```
